### TYPES OF DATA


*   Numerical Data
*   Categorical Data


Categorical data
Categorical data tells you which group something belongs to.

Examples from Seaborn's tips dataset:
- sex → Male, Female
- smoker → Yes, No
- day → Thur, Fri, Sat, Sun
- time → Lunch, Dinner
There are two common types:
- Nominal: Categories without a natural order. Example: sex, smoker status.
- Ordinal: Categories with a meaningful order. Example: Low, Medium, High.
2. Numerical data
Numerical data represents numbers that you can meaningfully calculate with.

Examples from the tips dataset:
- total_bill → Total bill amount
- tip → Tip amount
- size → Number of people at the table
Two common types:
- Discrete: Countable values, usually whole numbers. Example: number of people, number of orders.
- Continuous: Measurable values that can take decimal values. Example: height, weight, temperature.

A **multiple bar chart** (or grouped bar chart) is used to compare **multiple data series side-by-side** across the same categories.

In this specific code, it allows us to easily compare each **batsman's performance** across three different years (**2015, 2016, and 2017**) in a single visual.

---

 How the Code Works

* **Base Positions (`np.arange(df.shape[0])`):** Generates evenly spaced numeric positions (`0, 1, 2, ...`) on the x-axis for each batsman.
* **Bar Width (`width=0.2`):** Makes each bar narrow enough so that three bars can sit next to each other without overlapping.
* **Horizontal Offsets (`- 0.2` and `+ 0.2`):**
  * Shifts the **2015** bars slightly to the **left** (`- 0.2`).
  * Keeps the **2016** bars in the **center**.
  * Shifts the **2017** bars slightly to the **right** (`+ 0.2`).
* **Color Coding (`color`):** Assigns distinct colors (`yellow`, `red`, `blue`) to visually separate each year.
* **Category Labels (`plt.xticks`):** Replaces the numeric x-axis positions with the actual names from `df['batsman']`, centered under the middle bar.


```python
# printing multiple bar chart
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2, color='yellow')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2, color='red')
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width=0.2, color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

### STACKED BAR CHART

A stacked bar chart is a type of bar chart in which multiple categories are stacked on top of each other (vertical) or placed end-to-end (horizontal) to show how different parts contribute to a total.

In simple words, it helps you compare both the total and its individual components across different groups.

```python
#stacked bar chart
df
```

```python
plt.bar(df['batsman'], df['2015'],label='2015')
plt.bar(df['batsman'], df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'], df['2017'],bottom=df['2015']+df['2016'],label='2017')
plt.legend()
plt.show()
```

 What does `bottom` do in Matplotlib?

The `bottom` parameter tells Matplotlib **where to start drawing a bar vertically**.

- `bottom=df['2015']` means start the 2016 bar from the height of the 2015 bar.
- So, 2016 is placed on top of 2015.

- `bottom=df['2015'] + df['2016']` means start the 2017 bar after adding the heights of 2015 and 2016.
- So, 2017 is placed on top of both previous years.

### Simple Example

Suppose a batsman scored:

- 2015 = 100 runs
- 2016 = 150 runs
- 2017 = 200 runs

| Year | Starting position (`bottom`) |
|---|---:|
| 2015 | 0 |
| 2016 | 100 |
| 2017 | 250 (100 + 150) |

**Remember:**
- `bottom` decides the starting position of a bar.
- `df['2015']` means start after 2015.
- `df['2015'] + df['2016']` means start after 2015 and 2016 combined.

That is why we keep adding the previous years in `bottom`.

## PIE CHART

What is a Pie Chart in Matplotlib?

A pie chart is a circular graph divided into different slices. Each slice represents a category's contribution to the total value.

In simple words, it shows how a total is divided among different categories.

Why do we use a Pie Chart?
We use pie charts to:
1. Understand how a total is divided.
2. Compare the contribution of different categories.
3. Identify which category has the largest or smallest share.
4. Represent percentages visually.
5. Present data in an easy-to-understand format.

What type of data is used in a Pie Chart?

Pie charts are mainly used for categorical data with numerical values representing each category.

mainly for :


*   univariate / bivariate analysis
*   categorical vs numerical



```python
data = [23,45,100,29,65]
subjects=['eng','science','sst','maths','punjabi']
plt.pie(data,labels=subjects)
```

**Explode Function**

The explode parameter is used to separate or pull out one or more slices of a pie chart from the centre.

In simple words, it moves a selected slice slightly away from the pie chart to highlight it.

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%1.1f%%',explode=[0.1,0,0,0,0,0],shadow=True)
plt.show()
```

Here:
explode=[0.1, 0, 0, 0]

- 0.1 → Pulls the first slice (Rent) outward.
- 0 → Keeps the second slice (Food) in its normal position.
- 0 → Keeps the third slice in its normal position.
- 0 → Keeps the fourth slice in its normal position.

How does it work?

The values inside explode represent how far each slice moves away from the centre.

Explode value	Meaning

0 	No separation

0.1	Small separation

0.2	More separation

0.5	Large separation



---


Important points
- The number of values in explode should match the number of pie slices.
- Each value controls the separation of its corresponding slice.
- You can explode multiple slices at the same time.
Remember: explode is used to highlight an important slice by moving it away from the centre of the pie chart. It does not change the actual data or percentage.
