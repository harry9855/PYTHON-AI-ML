### TYPES OF DATA


*   Numerical Data
*   Categorical Data


Categorical data
Categorical data tells you which group something belongs to.

Examples from Seaborn's tips dataset:
- sex → Male, Female
- smoker → Yes, No
- day → Thur, Fri, Sat, Sun
- time → Lunch, Dinner
There are two common types:
- Nominal: Categories without a natural order. Example: sex, smoker status.
- Ordinal: Categories with a meaningful order. Example: Low, Medium, High.
2. Numerical data
Numerical data represents numbers that you can meaningfully calculate with.

Examples from the tips dataset:
- total_bill → Total bill amount
- tip → Tip amount
- size → Number of people at the table
Two common types:
- Discrete: Countable values, usually whole numbers. Example: number of people, number of orders.
- Continuous: Measurable values that can take decimal values. Example: height, weight, temperature.

```python
from matplotlib import colors
# Changing the colors
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Years')
plt.ylabel('Runs')
```

```python
# changing line style
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Years')
plt.ylabel('Runs')
```

```python
# marking marker at each point
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',marker='o')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Years')
plt.ylabel('Runs')
```

```python
# using legend and making label for each line ( basically which line belongs to whom)
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',marker='D', label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',marker='o',label='Rohit')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.legend() # (loc='best') default parameter which figure automatically empty face
# for plt.legend(loc='upper right') , 'right' , 'left' etc
```

### LIMITING AXES

Limiting axes in Matplotlib means setting the minimum and maximum values displayed on the X-axis and Y-axis of a plot.

It allows you to control the visible portion of a graph.

```python
#limiting axes y axis and x axis
price=[48000,50000,60000,65000,59000,57000,1000000]
year=[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
plt.ylim(0,100000)
plt.xlim(2015,2019)
```

### Grid Function

The grid() function in Matplotlib is used to display horizontal and vertical grid lines in a plot.

These lines make it easier to read values and understand the position of data points.

```python
# grid ( provides grid view for better understanding)
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',marker='D', label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',marker='o',label='Rohit')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.legend()
plt.grid()
```

### SHOW FUNCTION

In Matplotlib, plt.show() displays the graph or figure you have created. It tells Matplotlib that you want to render the plot so you can see it.

Google Colab uses IPython/Jupyter Notebook, which has a feature called automatic display.

```python
# Show function
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',marker='D', label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',marker='o',label='Rohit')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.legend()
plt.grid()
plt.show()
```

## SCATTER PLOT

A scatter plot is a type of graph used to show the relationship between two numerical variables.

It displays data as individual dots (points), where:
- X-axis represents the first variable.
- Y-axis represents the second variable.
- Each dot represents one observation or data point.
For example, suppose we want to understand whether studying more hours leads to higher marks.

It is used to study :


*   Bivariate Analysis : Bivariate analysis is the process of analysing the relationship between two variables to understand how they are connected or whether one changes when the other changes.
*   Numerical to Numerical analyis :It means analysing the relationship between two numerical variables to understand whether changes in one variable are associated with changes in the other.



2D plot is the special case for scatter plot

if we join scatter plot with line it becomes 2d plot

```python
# scatter plot example
x=np.linspace(-10,10,50)
#y=nx+b( equation for straight line )
y=10*x+1+np.random.randint(0,300,50) # using random values to print random scatter
plt.scatter(x,y)
```
