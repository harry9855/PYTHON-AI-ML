#SERIES

###Series from list

Series has basically two things 'indexes' which are genrated automatically and values which are given to it

````python
# String Series
country=["India", "USA", "China", "Russia", "Japan"]
pd.Series(country)
#It has dtype as object and 99% of times pandas says string as object
````

````python
# Integr Series
runs=[98,88,75,93,52,71,38,12,43]
runs=pd.Series(runs)
# It has dtype as int64
````

````python
# Creating Custom Indexes
marks_list=[98,92,78,85,90]
subjects=["Social Sci","Maths","Englis","Science","Punjabi"]
pd.Series(marks_list,index=subjects)
# One thing length indexes and length values should match
````

````python
# setiing a name
marks=pd.Series(marks_list,index=subjects,name="Hardik Marks")
marks
# In pd.Series, the name parameter gives a name/title to the Series.
# marks created here is a object of series
````

````python
print(marks)
````

###Series from dict

````python
marks_dict={"Social Sci":98,"Maths":92,"Englis":78,"Science":85,"Punjabi":90}
pd.Series(marks_dict,name="Hardik Marks")
# using dicitionary helps to convert keys to indexes and values to values in series
````

# MATHEMATICAL METHODS IN SERIES

some of the most used function are :

- `count()` : is used to count number of values in sereis similar to size function but one difference that count does not include null (NaN) values. It ignores missing values.

````python
vk.count()
````

- `sum() and product()`: It find sum and product the values in the series

````python
subs.sum()
````

````python
subs.product()
# answer is zero because it has any one element zero in the series
````

- `mean()` : Mean is the average of all values.

```text
Mean = Sum of all values / Total number of values

````python
print(vk.mean())
# it gives average run scored by virat kolhi
````

- `median()` is used to find the middle value of data after sorting. It automatically ignores NaN values.


```text
for odd values
Median = ((n + 1) / 2)th value
for even value
Median = Average of middle two values

````python
vk.median()
````

- `mode()` is used to find the most frequently occurring value in data. The value with highest frequency is called Mode.

````python
vk.mode()
````

- `std()` is used to calculate the Standard Deviation of data.

    Standard Deviation measures: how spread out values are from the mean
    
---

```text
σ = √( Σ(x - μ)² / N )

σ = standard deviation
x = each value
μ = mean
N = total number of values

````python
subs.std()
````

`var()` is used to calculate the Variance of data.

Variance measures:
- how far data values are spread from the mean

---



```text
σ² = Σ(x - μ)² / N

Where:

σ² = population variance
x = each value
μ = population mean
N = total number of observations

````python
subs.var()
````

`min()/max()` function are used to calculate minimum and maximum value in dataset

````python
subs.min()
````

````python
subs.max()
````

````python
# describe() it is a function used to describe some mathematical methods
subs.describe()
````
