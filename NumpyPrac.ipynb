#RULES FOR BROADCASTING


1. Make the two arrays have the same number of dimensions

* If the number of dimensions of the two arrays are different, add new dimensions of size 1 to the beginning (head) of the array with fewer dimensions.

Example : (3,3,3) this 3D array and (3) this is 1D array we need to make a smaller dimension array to always bigger dimension array (1,1,3)

2. Make each dimension of the two arrays the same size


*   If the sizes of the dimensions do not match, dimensions with size 1 are stretched to match the size of the other array.
*   If a dimension has different sizes in both arrays and neither of them is 1, then broadcasting cannot be performed, and an error is raised.

Example:if 3D array is of size (3,3,3) and new array which created is (1,1,3) then we have to convert it in (3,3,3) but we can only stretch or change the size of array which has dimension 1

actually it all happened internally b is converted from [0,1,2] to ([0,1,2]

#MEAN SQUARE ERROR
Formula for mean square error is :
$$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $$

Where:

$$ n = \text{total number of values} $$  
$$ y_i = \text{actual values} $$  
$$ \hat{y}_i = \text{predicted values} $$

so basically this formula calculates square of difference of actual values and predicted values and then add all the squared values and then fnd the average of that value

````python
actual=np.random.randint(1,50,25)
````

````python
predicted=np.random.randint(1,50,25)
````

````python
actual
````

````python
predicted
````

````python
def mse(actual,predicted):
  return np.mean((actual-predicted)**2)
````
