## SELECTING ROWS FROM DATAFRAME

We use Two Methods :


*   iloc : used to access data by its integer position
*   loc  : used to access data by its index/label



```python
# single row using iloc
movies.iloc[0]

# It is used fetch single row from movies
```

```python
# Multiple Rows using iloc
movies.iloc[5:7] # we can minpluate number inside like
# wants movies with alternate we will give [::2] lly we anything
```

```python
# multiple rows using fancy indexes
movies.iloc[[4,5,9]]
```

```python
students
```

## FILTERING DATAFRAMES

```python
# Find all winner of ipl
mask=ipl["MatchNumber"]=="Final"
new_df=ipl[mask]
new_df[["Season","WinningTeam"]]
```

```python
# how many super over have occured
mask=ipl["SuperOver"]=="Y"
ipl[mask].shape[0]
```

```python
# how many matches csk won in kolkata
ipl[(ipl["WinningTeam"]=="Chennai Super Kings")& (ipl["City"]=="Kolkata")].shape[0]

```

```python
# Toss winner is a match winner percentage
mask=ipl["TossWinner"]==ipl["WinningTeam"]
count=ipl[mask].shape[0]
total=ipl.shape[0]
perc=count/total*100
perc
```

```python
# Movies with rating higher than 8 and votes>10000
movies[(movies["imdb_rating"]>8) & (movies["imdb_votes"]>10000)].shape[0]
```

```python
# Action movies with rating greater than 7.5
movies[(movies["genres"].str.contains("Action")) & (movies["imdb_rating"]>7.5)].shape[0]

# we can also use apply , apply function is used to apply any specific function
#movies[(movies["genres"].str.split('|').apply(lambda x:"Action" in x)) & (movies["imdb_rating"]>7.5)].shape[0]
#
```
