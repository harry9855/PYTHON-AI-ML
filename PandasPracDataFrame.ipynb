# GROUP BY

## groupby()
groupby() in Pandas is used when you want to divide your data into groups based on a column and then perform an operation on each group

Basically we form groups on the basis of column and there are commonly two types of column :

1.   Numerical Column
2.   Categorical Column



````python
movies_imdb=pd.read_csv('imdb-top-1000.csv')
````

````python
genres=movies_imdb.groupby('Genre')
````

````python
genres.sum()
````

`groupby()` is used to **divide the data into different groups based on a particular column**.

For example, suppose we have a movie dataset with a `Genre` column.

```python
Genre
------
Action
Comedy
Action
Drama
Comedy
Action
```

If we use:

```python
genre = movies.groupby('Genre')
```

Pandas creates groups based on the **distinct values of the `Genre` column**.

Conceptually, the data is divided into:

```text
Action Group
→ All movies whose Genre is Action

Comedy Group
→ All movies whose Genre is Comedy

Drama Group
→ All movies whose Genre is Drama
```

So `groupby()` creates a **grouped object** containing these groups.

---

## Applying an Operation After GroupBy

Now suppose we use:

```python
genre.sum()
```

Here:

```text
groupby('Genre')
        ↓
Creates groups based on Genre
        ↓
sum()
        ↓
Calculates the sum separately for each Genre
```

For example:

```python
genre = movies.groupby('Genre')
genre.sum()
```

The result will contain one result for each genre.

Conceptually:

```text
Action  → Sum of values belonging to Action movies
Comedy  → Sum of values belonging to Comedy movies
Drama   → Sum of values belonging to Drama movies
```

---

# Important Concept

`groupby()` itself **does not calculate anything**.

It only creates groups.

The actual calculation happens when we apply an operation such as:

```python
sum()
mean()
count()
max()
min()
median()
```

For example:

```python
genre.sum()
```

means:

> Group the data according to `Genre`, and then calculate the sum separately for each group.

---

# Example

Suppose we have:

```text
Genre      Budget
Action     100
Comedy      50
Action     200
Comedy      70
Drama      150
```

After:

```python
genre = movies.groupby('Genre')
```

the data is conceptually divided into:

```text
Action
100
200

Comedy
50
70

Drama
150
```

Now:

```python
genre.sum()
```

gives:

```text
Action    → 300
Comedy    → 120
Drama     → 150
```

Because:

```text
Action  → 100 + 200 = 300
Comedy  → 50 + 70 = 120
Drama   → 150
```

---

# GroupBy Formula

The basic idea is:

```text
Group → Apply Operation → Result
```

For example:

```python
movies.groupby('Genre').sum()
```

means:

```text
Group movies by Genre
        ↓
Create separate groups for each Genre
        ↓
Apply sum()
        ↓
Get the sum for each Genre
```

---

# Important Note

It is better to think of `groupby()` as creating **logical groups**, rather than literally creating separate DataFrames.

For example:

```python
genre = movies.groupby('Genre')
```

does not immediately create several independent tables.

It creates a **GroupBy object** that knows how the rows are divided into groups.

When you call:

```python
genre.sum()
```

or:

```python
genre.mean()
```

Pandas performs the operation group-by-group and returns the result.

---

# Easy Way to Remember

```text
groupby() → Creates groups
sum()     → Adds values within each group
mean()    → Calculates average within each group
count()   → Counts values within each group
max()     → Finds maximum within each group
min()     → Finds minimum within each group
```

Therefore:

```python
genre = movies.groupby('Genre')
genre.sum()
```

means:

> **Create groups based on Genre → perform sum separately for every Genre.**

# MERGING , JOINING AND CONCATENATING

## Reading All CSV

````python
courses=pd.read_csv("/content/courses.csv")
delivery=pd.read_csv("/content/deliveries1.csv")
matches=pd.read_csv("/content/matches.csv")
nov=pd.read_csv("/content/reg-month1.csv")
dec=pd.read_csv("/content/reg-month2.csv")
students_new=pd.read_csv("/content/students.csv")
````

## CONCAT


````python
# Concatenation
#pd.concat is a function that is used to join to dataframe together vertically
regs=pd.concat([nov,dec],ignore_index=True) # ignore_index is used to ignore default of indexes of each df
# and apply new indexes from start to combined dataframe(df)
regs
````

````python
# If we want our Orignal indexes of the df we can simply combine two df
pd.concat([nov,dec])
# Here we can see indexes are repeating so it will be difficult to distinguish
````

````python
# so to save our previous indexes we need multi indexing that is providing a index to specific df
multi=pd.concat([nov,dec],keys=['Nov','Dec'])
multi
````

````python
# As we can see two indexes are made for each data frame a one index and orignal index inside a df
# if we want to navigate we can use iloc
multi.loc[('Nov')] # it gives df
multi.loc[('Nov',0)] # it provides specific data from that df
````

````python
# We can also stack df horizontally
pd.concat([nov,dec],axis=1)
````

## MERGE

````python
# Join
# joins are used to combine data from two DataFrames based on a common column or index.
````

````python
# Inner Join
# An inner join combines two DataFrames and keeps only the rows where the common
# column has a matching value in both DataFrames.
````

````python
students_new.merge(regs,how='inner',on='student_id')
````

````python
# left join
# it also shows common things but also all the data of left table
courses.merge(regs,how='left',on='course_id')
````

````python
# right join
# similar as left join but it shows data of right table
````

````python
students_new.merge(regs,how='right',on='student_id')
````

````python
# outer join
# it gives full data of both table
````

````python
students_new.merge(regs,how='outer',on='student_id')
````
