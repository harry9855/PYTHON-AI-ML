## apply FUNCTION

```python
# apply function is both series and df function
# it is used to apply a sepcific function on all values of series
temp1=pd.Series([100,-200,500,-100])
```

```python
def sigmoid(value):
  return 1/1+np.exp(-value)
```

```python
# now if we want to apply sigmoid function to each value of the series then we use apply
temp1.apply(sigmoid)
```

```python
# apply function on dataframe
points_df = pd.DataFrame(
    {
        '1st point': [(3,4), (-6,5), (0,0), (-10,1), (4,5)],
        '2nd point': [(-3,4), (0,0), (2,2), (10,10), (1,1)]
    }
)

points_df
```

```python
# now if we want calculate euclidean distance of 1st and 2nd point
def euclidean(row):
  pt_a=row['1st point'] # name of the col should be correct
  pt_b=row['2nd point'] # name of the col should be correct
  return ((pt_a[0]-pt_b[0])**2+(pt_a[1]-pt_b[1])**2)**0.5
# here pt_a will store 1 st point and pt_b will store 2 nd point
# because we will be passing particular row to function instead col as we do in series
```

```python
points_df.apply(euclidean,axis=1) # we have made axis=1 so that data goes row wise
# because default is column
```
