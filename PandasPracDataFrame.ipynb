## FILTERING DATAFRAMES

```python
# Find all winner of ipl
mask=ipl["MatchNumber"]=="Final"
new_df=ipl[mask]
new_df[["Season","WinningTeam"]]
```

```python
# how many super over have occured
mask=ipl["SuperOver"]=="Y"
ipl[mask].shape[0]
```

```python
# how many matches csk won in kolkata
ipl[(ipl["WinningTeam"]=="Chennai Super Kings")& (ipl["City"]=="Kolkata")].shape[0]

```

```python
# Toss winner is a match winner percentage
mask=ipl["TossWinner"]==ipl["WinningTeam"]
count=ipl[mask].shape[0]
total=ipl.shape[0]
perc=count/total*100
perc
```

```python
# Movies with rating higher than 8 and votes>10000
movies[(movies["imdb_rating"]>8) & (movies["imdb_votes"]>10000)].shape[0]
```

```python
# Action movies with rating greater than 7.5
movies[(movies["genres"].str.contains("Action")) & (movies["imdb_rating"]>7.5)].shape[0]

# we can also use apply , apply function is used to apply any specific function
#movies[(movies["genres"].str.split('|').apply(lambda x:"Action" in x)) & (movies["imdb_rating"]>7.5)].shape[0]
#
```

## ADDING NEW COLS

```python
# adding col
movies["Country"]="India"
movies.head()
```

```python
# As we can see id is written as int 64 we can change it to int32
ipl['ID']=ipl['ID'].astype('int32')
```

```python
ipl.info()
# here we can see memory usage has been reduced
# basically category is a type of data which is to be repeated like enums
# category is a Pandas datatype used for columns that contain a limited number of repeated values
```

```python
ipl['Team1']=ipl['Team1'].astype('category')
ipl['Team2']=ipl['Team2'].astype('category')
ipl['WinningTeam']=ipl['WinningTeam'].astype('category')
ipl['TossWinner']=ipl['TossWinner'].astype('category')
```

## value_counts FUNCTION

```python
marks_dict={
    "IQ":[100,90,120,80,80],
    "MARKS":[90,70,100,70,70],
    "PACKAGE":[10,7,14,20,20]
}
marks=pd.DataFrame(marks_dict)
```

```python
# Value_counts() it is a function that calculates the frequency of rows in the dataframe
marks.value_counts()
```

```python
# find player which has won most man of the match in final and qualifiers
ipl[~ipl["MatchNumber"].str.isdigit()]["Player_of_Match"].value_counts()
```

```python
# Toss decision Wether team choose to ball or bat after winning toss
ipl['TossDecision'].value_counts()
```

```python
# how many matches each team has played
(ipl['Team1'].value_counts()+ipl['Team2'].value_counts()).sort_values(ascending=False)
```

## rank FUNCTION

```python
# rank() is a utility function only applicable on series not dataframes
runs['batsman_rank']=runs['batsman_run'].rank(ascending=False)
```

```python
runs.sort_values('batsman_rank')
```
