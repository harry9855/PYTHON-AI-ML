```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Colored Scatter Plot

```python
iris=pd.read_csv('/content/iris.csv')
iris.head()
```

```python
iris.sample(5)
```

```python
iris['Flower_Number'] = iris['Species'].map({
    'Iris-setosa': 1,
    'Iris-virginica': 2,
    'Iris-versicolor': 3
})
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Flower_Number'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.show()
```

A coloured scatter plot uses different colours to represent different groups or categories of data.


A normal scatter plot uses dots to show the relationship between two numerical variables. Adding colours gives us an additional piece of information about each dot.

Basiclly in above it tell different length 3 different flowers

In Matplotlib, both c and color are used to control colours, but they behave differently.

Parameter	Purpose

color	Sets a fixed colour for all points/bars/lines.
c	can set a fixed colour or assign colours based on numerical values.

Points get different colours based on their Flower_Number values.

```python
# cmap
# cmap decides which colours Matplotlib uses for your numerical values.
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Flower_Number'],cmap='Spectral')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.show()
```

```python
# colorbar() — Colour scale indicator
# It displays a scale beside your graph explaining which colour represents which numerical value.
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Flower_Number'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.show()
```

```python
# alpha — Transparency (OPACITY)
# alpha controls how transparent your points are.
# Its value ranges from 0 to 1
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Flower_Number'],alpha=0.6)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.show()
```

## PLOT SIZE

```python
plt.figure(figsize=(13,7)) # it is used to change the size of the graph
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Flower_Number'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.show()
```

## SUB PLOTS

A subplot allows you to display multiple graphs inside a single figure.

Instead of creating separate figures for each graph, you arrange them in rows and columns.

Why do we use subplots?

Suppose you want to analyse the Iris dataset using three graphs:
- Sepal Length vs Petal Length
- Sepal Width vs Petal Width
- Sepal Length distribution

You can display all three graphs together using subplots.

```python
fig,ax=plt.subplots(figsize=(6,5))
# when we run plt.subplot it gives two object figure object and axis object that we stored in fig and ax
ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title('BATSMAN STRIKE RATE AND AVERAGE')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')

fig.show() # instead of plt.show()
```

fig and ax are variables that store two important objects:
- fig → Figure object (the complete canvas).
- ax → Axes object (the actual plotting area where your graph is drawn).

Figure object (fig)

A Figure is the entire area or canvas on which you create your plots.

It controls things like:
- Overall figure size.
- Background.
- Multiple subplots.
- Saving the complete figure.

Axes object (ax)

An Axes object represents an individual plotting area inside the figure.

It controls:
- X-axis and Y-axis.
- Labels and titles.
- Scatter plots, histograms, bar charts.
- Grid and limits.|

```python
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True) # sharex is used if both graph use same x label
# due to which last graph will show the cordinates of x axis
# when we run plt.subplot it gives two object figure object and axis object that we stored in fig and ax
ax[0].scatter(sample_df['avg'],sample_df['strike_rate'])
ax[0].set_title('BATSMAN STRIKE RATE AND AVERAGE')
ax[0].set_ylabel('Strike Rate')

ax[1].scatter(sample_df['avg'],sample_df['runs'])
ax[1].set_title('BATSMAN RUNS AND AVERAGE')
ax[1].set_xlabel('Average')
ax[1].set_ylabel('runs')

fig.show() # instead of plt.show()
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(13,10))
# here axis becomes 2d object
ax[0][0].scatter(sample_df['avg'],sample_df['strike_rate'])
ax[0][0].set_title('BATSMAN STRIKE RATE AND AVERAGE')
ax[0][0].set_ylabel('Strike Rate')

ax[1][0].scatter(sample_df['avg'],sample_df['runs'])
ax[1][0].set_title('BATSMAN RUNS AND AVERAGE')
ax[1][0].set_xlabel('Average')
ax[1][0].set_ylabel('runs')


ax[0][1].hist(sample_df['runs'],log=True)
ax[0][1].set_title('BATSMAN RUNS')
ax[0][1].set_xlabel('Average')
ax[0][1].set_ylabel('Number Of Batsman')

ax[1][1].hist(sample_df['avg'])
ax[1][1].set_title('BATSMAN AVERAGE')
ax[1][1].set_xlabel('Average')
ax[1][1].set_ylabel('Number Of Batsman')
fig.show() # instead of plt.show()
```

## 3D PLOTS

A 3D plot (three-dimensional plot) is a graph that represents data using three dimensions: X, Y and Z axes.

A normal scatter plot uses two variables (X and Y). A 3D scatter plot uses three numerical variables.

For example, imagine analysing students:
* **X** = Study hours
* **Y** = Sleep hours
* **Z** = Exam marks

A 3D plot helps you observe how all three variables relate to each other. We use them when we want to visualise relationships between three numerical variables.

 **Common Applications**

| Field | X, Y and Z variables |
| :--- | :--- |
| **Education** | Study hours, sleep hours, marks |
| **Finance** | Risk, return, investment amount |
| **Real estate** | House area, number of rooms, price |
| **Engineering** | Voltage, current, resistance |
| **Machine Learning** | Feature 1, Feature 2, predicted value |

---

**Where is 3D plotting required?**

You may encounter it in:
1. **Exploratory Data Analysis (EDA):** Exploring relationships among three numerical features.
2. **Machine Learning:** Visualising three-dimensional feature spaces or model predictions.
3. **Engineering:** Analysing physical systems, surfaces and simulations.
4. **Finance:** Exploring relationships among multiple financial variables.
5. **Scientific research:** Visualising spatial data and experimental results.

---

**One important limitation**

> **Note:** Don't assume 3D is always better than 2D.

3D graphs can become difficult to read because:
* Points may overlap.
* Some points may hide behind others.
* Perspective can make distances difficult to compare.


### 3D SCATTER PLOT

A 3D scatter plot is a graph that displays individual data points using three numerical variables: X, Y and Z.

In a normal 2D scatter plot, each dot has two coordinates (X, Y). In a 3D scatter plot, each dot has three coordinates (X, Y, Z).

When is it useful?
1. Comparing three numerical variables simultaneously.
2. Finding clusters or groups in data.
3. Identifying patterns and outliers.
4. Exploring three-feature datasets in machine learning.
>Remember:
- 2D scatter plot = two numerical variables.
- 3D scatter plot = three numerical variables.
- Each dot = one observation (one flower in your Iris dataset).

```python
fig=plt.figure(figsize=(13,10))
ax=plt.axes(projection='3d')
ax.scatter(sample_df['avg'],sample_df['strike_rate'],sample_df['runs'])
ax.set_title('BATSMAN ANALYSIS')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')
ax.set_zlabel('Runs')
```

### 3D LINE PLOT

A 3D line plot is a graph that connects data points using lines in three dimensions: X, Y and Z.

Unlike a 3D scatter plot, which displays individual dots, a 3D line plot connects the points to show a continuous path or progression.

**Why do we use it?**

A 3D line plot is useful when you want to understand how data changes along a sequence while considering three variables.

#### Examples

| Field | X | Y | Z |
| :--- | :--- | :--- | :--- |
| **Engineering** | Time | Voltage | Current |
| **Finance** | Time | Stock price | Trading volume |
| **Physics** | X position | Y position | Z position |
| **Machine learning** | Parameter 1 | Parameter 2 | Model loss |

```python
x=[0,12,35]
y=[0,6,21]
z=[0,17,29]
fig=plt.figure(figsize=(13,10))
ax=plt.axes(projection='3d')
ax.scatter(x,y,z,s=[100,100,100])
ax.plot3D(x,y,z,color='green')
```

### SURFACE PLOT

A **3D surface plot** is a three-dimensional graph that displays data as a continuous surface rather than individual dots or connected lines.

It uses three variables:
* **X-axis:** First input variable.
* **Y-axis:** Second input variable.
* **Z-axis:** Output value or height of the surface.


Think of it like a landscape:
* **Mountains** represent higher Z values.
* **Valleys** represent lower Z values.
* **Flat areas** represent relatively constant Z values.

---

**Why do we use it?**

A 3D surface plot helps us understand how an output changes when two input variables change simultaneously.

>examples

| Field | X-axis | Y-axis | Z-axis |
| :--- | :--- | :--- | :--- |
| **Engineering** | Voltage | Current | Power |
| **Finance** | Risk | Investment period | Expected return |
| **Machine Learning** | Parameter 1 | Parameter 2 | Model loss |
| **Physics** | Position X | Position Y | Temperature |


**Understand the important functions**

| Function | Purpose |
| :--- | :--- |
| `np.linspace()` | Generates evenly spaced values |
| `np.meshgrid()` | Creates a 2D coordinate grid from X and Y |
| `ax.plot_surface()` | Creates the 3D surface |
| `cmap='viridis'` | Controls the surface colour scheme |

> **What is `meshgrid()` doing?**
>
> It creates combinations of X and Y coordinates so that Matplotlib can calculate and draw the surface height (Z) at each coordinate.


**Where is it useful in Machine Learning?**

One important application is visualising a model's loss function.

Imagine:
* **X** = Learning rate
* **Y** = Model complexity
* **Z** = Model loss

A surface plot can help you see which combinations of parameters produce lower loss.

> **Important:** A surface plot generally requires structured X-Y grid data. You cannot simply pass three unrelated DataFrame columns to `plot_surface()` as you do with `scatter()`.

> **Remember:** **Scatter** = dots, **Line** = connected points, **Surface** = continuous 3D shape.


```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
X,Y=np.meshgrid(x,y)
Z=X**2+Y**2
fig=plt.figure()
ax=plt.axes(projection='3d')
p=ax.plot_surface(X,Y,Z,cmap='viridis')
fig.colorbar(p)
```

```python
# creating other
Z=np.sin(X)+np.cos(Y)
fig=plt.figure()
ax=plt.axes(projection='3d')
p=ax.plot_surface(X,Y,Z,cmap='viridis')
fig.colorbar(p)
```

### CONTOUR PLOT

A **contour plot** is a 2D visualization used to represent a 3D surface (or a function of two variables) on a flat 2D graph.

It shows how the value of a function changes with two input variables, using lines called **contour lines**.

---
**Understand it with an example**

Imagine a hill. Its height depends on your position:
* **X-axis** = horizontal position
* **Y-axis** = vertical position
* **Z-axis** = height of the hill

A 3D surface plot shows the hill in three dimensions.

A contour plot shows the same hill from directly above, like looking at a map.


**What are contour lines?**

Each contour line connects points that have the same Z-value.

For example, suppose:
$$f(x,y) = x^2 + y^2$$

* **Contour level 1:** all points where $x^2 + y^2 = 1$
* **Contour level 4:** all points where $x^2 + y^2 = 4$
* **Contour level 9:** all points where $x^2 + y^2 = 9$


**What is `contourf()`?**

Matplotlib provides two commonly used functions:

| Function | Purpose |
| :--- | :--- |
| `plt.contour()` | Draws contour lines |
| `plt.contourf()` | Fills the regions between contour levels with colours |

#### Important parameters
* **`X`, `Y`:** Coordinate grids.
* **`Z`:** Function values at each coordinate.
* **`cmap`:** Colour scheme.
* **`levels`:** Number or specific values of contour levels.
* **`colorbar()`:** Displays what each colour represents.

---

**Why are contour plots used?**

1. **Geography:** Show elevation, mountains and valleys on topographic maps.
2. **Weather:** Show temperature, pressure and rainfall distribution.
3. **Engineering:** Show stress, heat and pressure distribution.
4. **Machine Learning:** Visualize loss functions and optimization.
5. **Mathematics:** Understand functions involving two variables.


```python
fig=plt.figure()
ax=plt.axes()
Z=X**2+Y**2
p=ax.contour(X,Y,Z,cmap='viridis')
fig.colorbar(p)
```

```python
fig=plt.figure()
ax=plt.axes()
Z=X**2+Y**2
p=ax.contourf(X,Y,Z,cmap='viridis')
fig.colorbar(p)
```

## HEATMAP

A heatmap is a data visualization that uses colours to represent numerical values. It helps you quickly identify high values, low values, patterns and relationships in a dataset.

Think of it as a table where each cell has a different colour depending on its value.

```python
delivery=pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022.csv')
```
